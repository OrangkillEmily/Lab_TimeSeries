```python
# BÀI 2: KIỂM ĐỊNH ADF TRÊN CHUỖI GỐC

import pandas as pd
from statsmodels.tsa.stattools import adfuller

# Đọc dữ liệu
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(url, parse_dates=["Month"], index_col="Month")

# Lấy chuỗi số hành khách gốc
series = df["Passengers"]

# Kiểm định ADF
result = adfuller(series, regression="c", autolag="AIC")

# In kết quả
print("===== KIỂM ĐỊNH ADF TRÊN CHUỖI GỐC =====")
print("Số quan sát:", len(series))
print("ADF statistic:", result[0])
print("p-value:", result[1])
print("Số độ trễ:", result[2])
print("Số quan sát sử dụng:", result[3])

print("\nGiá trị tới hạn:")
for level, value in result[4].items():
    print(f"{level}: {value:.4f}")

# Trả lời câu hỏi và kết luận
print("\n===== KẾT LUẬN =====")
print("H0: Chuỗi có nghiệm đơn vị, tức là không dừng.")
print("H1: Chuỗi không có nghiệm đơn vị.")

if result[1] < 0.05:
    print("p-value < 0.05: Bác bỏ H0.")
    print("Kết luận: Có bằng chứng chuỗi dừng.")
else:
    print("p-value >= 0.05: Chưa đủ bằng chứng bác bỏ H0.")
    print("Kết luận: Chuỗi gốc chưa dừng.")
```
